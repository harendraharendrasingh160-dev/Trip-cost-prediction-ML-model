````python
# Merge datasets into master dataframe
travel_df = flights_df.merge(users_df, left_on='userCode', right_on='code', how='inner')
travel_df = travel_df.merge(hotels_df, on='travelCode', how='left', suffixes=('_flight', '_hotel'))
````

````python
# Make new column
travel_df['total_trip_cost'] = travel_df['total_flight_cost'].fillna(0) + travel_df['total_hotel_cost'].fillna(0)
travel_df['one_day_trip_cost'] = travel_df['total_flight_cost'].fillna(0) + travel_df['price_hotel'].fillna(0)
````

````python
# Filling numerical columns null value
travel_df['days'] = travel_df['days'].fillna(0)
travel_df['price_hotel'] = travel_df['price_hotel'].fillna(0)
travel_df['total_hotel_cost'] = travel_df['total_hotel_cost'].fillna(0)

# Filling categorical columns null value
travel_df['name_hotel'] = travel_df['name_hotel'].fillna('No_Hotel')
travel_df['place'] = travel_df['place'].fillna('No_Hotel')
travel_df['date_hotel'] = travel_df['date_hotel'].fillna('None')
````

````python
# Replace value
travel_df["gender"]=travel_df["gender"].replace("none","Unknown")
````

````python
# Remove white space from both side for MAKE dataset easier to compare
travel_df.columns = travel_df.columns.str.strip()
string_cols = travel_df.select_dtypes(include=['object', 'category']).columns

for col in string_cols:
    travel_df[col] = travel_df[col].astype(str).str.strip()

travel_df.replace('nan', np.nan, inplace=True)
````

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

### Hypothetical Statement - 1 - Gender vs. Total Trip Spending

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

$H_0$ (Null Hypothesis): The mean total trip spending for male users is equal to the mean total trip spending for female users ($\mu_{\text{male}} = \mu_{\text{female}}$).  

$H_1$ (Alternate Hypothesis): The mean total trip spending differs significantly between male and female users ($\mu_{\text{male}} \neq \mu_{\text{female}}$).

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
male_spend = travel_df[travel_df['gender'] == 'Male']['total_trip_cost'].dropna()
female_spend = travel_df[travel_df['gender'] == 'Female']['total_trip_cost'].dropna()

t_stat, p_val_gender = stats.ttest_ind(male_spend, female_spend, equal_var=False)

print("=== TEST 1: Gender vs. Total Trip Spend ===")
print(f"Male Mean Spend  : ${male_spend.mean():.2f}")
print(f"Female Mean Spend: ${female_spend.mean():.2f}")
print(f"t-statistic      : {t_stat:.4f}")
print(f"p-value          : {p_val_gender:.4f}")
print("Result           : Fail to reject H0 (No significant difference)\n")
````

##### Which statistical test have you done to obtain P-Value?

Welch's Two-Sample Independent t-test.


##### Why did you choose the specific statistical test?

Welch's t-test compares the means of a continuous variable (total_trip_cost) across two independent categorical groups (gender) without requiring equal sample variances.

### Hypothetical Statement - 2 - Flight Class vs. Ticket Price

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

$H_0$ (Null Hypothesis): Mean ticket prices for First-Class/Premium flights are equal to Economic class flights ($\mu_{\text{first}} = \mu_{\text{economic}}$).  

$H_1$ (Alternate Hypothesis): Mean ticket prices for First-Class/Premium flights are significantly higher than Economic class flights ($\mu_{\text{first}} > \mu_{\text{economic}}$).

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
flight_class_groups = [group['total_flight_cost'].dropna() for _, group in travel_df.groupby('flightType')]
f_stat_class, p_val_class = stats.f_oneway(*flight_class_groups)

print("=== TEST 2: Flight Class vs. Ticket Price ===")
for f_type, group in travel_df.groupby('flightType'):
    price = group['total_flight_cost'].dropna()
    print(f"Class '{f_type:10s}': Mean = ${price.mean():.2f}")
print(f"F-statistic      : {f_stat_class:.4f}")
print(f"p-value          : {p_val_class:.4e}")
print("Result           : Reject H0 (Significant price differences between classes)\n")
````

##### Which statistical test have you done to obtain P-Value?

One-Way ANOVA test

##### Why did you choose the specific statistical test?

It evaluates whether the mean of a continuous variable (price) in one independent group is significantly greater than in another continuous distribution (flightType).

### Hypothetical Statement - 3 - Company vs. Flight Class Choice

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

$H_0$ (Null Hypothesis): Flight class choice (flightType) is independent of the user's corporate affiliation (company).

 $H_1$ (Alternate Hypothesis): Flight class choice (flightType) is significantly associated with the user's corporate affiliation (company).

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
contingency_table = pd.crosstab(travel_df['company'], travel_df['flightType'])
chi2_stat, p_val_chi2, dof, _ = stats.chi2_contingency(contingency_table)

print("=== TEST 3: Company vs. Flight Class Choice ===")
print("Contingency Table:")
print(contingency_table)
print(f"\nChi2 Statistic   : {chi2_stat:.4f}")
print(f"Degrees of Freedom: {dof}")
print(f"p-value          : {p_val_chi2:.4f}")
print("Result           : Reject H0 (Flight class choice depends on company)\n")
````

##### Which statistical test have you done to obtain P-Value?

Chi-Square ($\chi^2$) Test of Independence.

````python
# dealing with date columns

# 1) Convert strings to datetime objects
travel_df["date_flight"] = pd.to_datetime(travel_df["date_flight"], errors="coerce")
travel_df["date_hotel"] = pd.to_datetime(travel_df["date_hotel"], errors="coerce")

# 2) Extract numeric datetime components
for col in ["date_flight"]:
    travel_df[f"{col}_year"] = travel_df[col].dt.year
    travel_df[f"{col}_month"] = travel_df[col].dt.month
    travel_df[f"{col}_day"] = travel_df[col].dt.day
    travel_df[f"{col}_dayofweek"] = travel_df[col].dt.dayofweek
    travel_df[f"{col}_is_weekend"] = (
        travel_df[col].dt.dayofweek >= 5
    ).astype(int)

for col in ["date_hotel"]:
    travel_df[f"{col}_year"] = travel_df[col].dt.year.fillna(-1).astype(int)
    travel_df[f"{col}_month"] = travel_df[col].dt.month.fillna(-1).astype(int)
    travel_df[f"{col}_day"] = travel_df[col].dt.day.fillna(-1).astype(int)
    travel_df[f"{col}_dayofweek"] = travel_df[col].dt.dayofweek.fillna(-1).astype(int)
    travel_df[f"{col}_is_weekend"] = (
        travel_df[col].dt.dayofweek.fillna(-1) >= 5
    ).astype(int)
````

````python
# ML Model - 2 Implementation

# 1. Pipe creation

rf_pipeline = Pipeline(steps=[
    ('scaler', StandardScaler()),
    ('rf', RandomForestRegressor(random_state=42, n_jobs=-1))
])

# 2. Hyperparameter tunning
param_distributions = {
    'rf__n_estimators': [20, 30, 50],
    'rf__max_depth': [10, 12, 15],
    'rf__max_samples': [0.3, 0.5, 0.8],
    'rf__min_samples_split': [2, 5, 10],
    'rf__min_samples_leaf': [1, 2, 4]
}

random_search = RandomizedSearchCV(
    estimator=rf_pipeline,
    param_distributions=param_distributions,
    n_iter=5,
    cv=3,
    scoring='r2',
    random_state=42,
    n_jobs=-1,
    verbose=1
)

# 3. Fit algorithm

print("--- Starting Random Forest Hyperparameter Tuning & Fitting ---")
random_search.fit(X_train, y_train)
best_rf_pipeline = random_search.best_estimator_

# 4. Predict

y_pred_train = best_rf_pipeline.predict(X_train)
y_pred_test = best_rf_pipeline.predict(X_test)

# 5. Evaluation metric Score Chart

trf_mae = mean_absolute_error(y_test, y_pred_test)
trf_rmse = np.sqrt(mean_squared_error(y_test, y_pred_test))
trf_r2_train = r2_score(y_train, y_pred_train)
trf_r2_test = r2_score(y_test, y_pred_test)

print("\n--- BEST HYPERPARAMETERS FOUND ---")
print(random_search.best_params_)

print("\n--- TUNED RANDOM FOREST PERFORMANCE ---")
print(f"Train R² Score : {trf_r2_train:.4f}")
print(f"Test R² Score  : {trf_r2_test:.4f}")
print(f"Test MAE       : ${trf_mae:.2f}")
print(f"Test RMSE      : ${trf_rmse:.2f}")
````
